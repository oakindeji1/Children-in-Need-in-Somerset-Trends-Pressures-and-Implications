```python
#Created an explicit “geography level” flag
# region_code / region_name (13 blanks)
# These rows are England/ National total rows

import numpy as np

df["geo_level"] = np.where(
    df["la_name"].notna(), "Local authority",         #Checks if the "la_name" column has a non-null value. If true, assigns "Local authority" to that row.
    np.where(df["region_name"].notna(), "Region", "England")    #This is the nested condition that runs when "la_name" is null. It checks if "region_name" has a non-null value. If true, assigns "Region", otherwise assigns "England".
)
```

```python
df["region_name"] = df["region_name"].fillna("England") #This replaces any missing values in the "region_name" column with the string "England".
df["region_code"] = df["region_code"].fillna("ENG")  #This replaces any missing values in the "region_code" column with the string "ENG".

df["la_name"] = df["la_name"].fillna("Not applicable") # This replaces any missing values in the "la_name" column with the string "Not applicable".
df["old_la_code"] = df["old_la_code"].fillna("NA")
df["new_la_code"] = df["new_la_code"].fillna("NA")
```

```python
# Create numeric analysis columns
import numpy as np
SUPPRESSION_VALUES = ["c", "x", "z", "u"]
```

```python
def cin_numeric(series):
    return (
        series
        .replace(SUPPRESSION_VALUES, np.nan)
        .replace("k", 0)
        .astype(float)
    )
```

```python
def cin_numeric(series):
    return (
        series
        .replace(SUPPRESSION_VALUES, np.nan)
        .replace("k", 0)
        .astype(float)
    )
```

```python
#This prevents accidental misuse later
cin_columns = [
    "anypoint_child",
    "anypoint_episodes",
    "started_child",
    "started_episodes",
    "ended_child",
    "ended_episodes",
    "at31_episodes",
    "awaiting_assessment"
    
]

for col in cin_columns:
    df[col + "_num"] = cin_numeric(df[col])
```

```python
from sqlalchemy import create_engine
import urllib

server = "OLUGBEMIGA"      # must match SSMS
database = "Sommerset"
driver = "ODBC Driver 17 for SQL Server"

params = urllib.parse.quote_plus(
    f"DRIVER={{{driver}}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")
```

```python
df.to_sql("children_in_need", engine, if_exists="replace", index=False, schema="dbo", chunksize=1000)
print("Export complete")
```
